Title:

Song Suggestion


Overview:

The Song-Suggestion project is what the title implies, is a recommendation system using your current mood and the genre of music you are into to recommend you music. The project uses simple feature engineering techniques, thus providing proper user based recommendations. 


Background:

Music has always been something I am passionate about, and I notices platforms such as spotify have recommendation systems based on music you listen to, but not so much based on your current emotion. Along with that there are times I cannot listen to a whole song, and enjoy skipping through songs, that match the same "vibe". I will be using this kaggle dataset: https://www.kaggle.com/datasets/maharshipandya/-spotify-tracks-dataset, to get recommendations based on mood. Eventually I plan on using python built in mfcc ( Mel-frequency cepstral coefficients) functionalities, which is a machine learning process to extract audio frequency, to properly mix songs based on the recommendations. 

Data Sources:

For this project I will be using https://www.kaggle.com/datasets/maharshipandya/-spotify-tracks-dataset 


Exploratory Analysis:

This project uses the Spotify Tracks Dataset from Kaggle, https://www.kaggle.com/datasets/maharshipandya/-spotify-tracks-dataset, a publicly available dataset containing over 180,000 tracks. The dataset includes features such as track name, artist, genre, popularity, acousticness, danceability, energy, instrumentalness, tempo, and more. When working on my emotion based song recommendation system I will be using these selected_feautures = ['danceability', 'energy', 'valence', 'tempo', 'instrumentalness', 'acousticness', 'speechiness', 'popularity']

Product:

The end product is a simple web application where users can receive song recommendations based on genre's theyre interested in and their current emotion. I hope to also implement a mixer feature using the MFCC functions. The backend is developed using Python with frameworks like Flask to handle requests and serve recommendations. Machine learning models are implemented using libraries such as scikit-learn and TensorFlow.
